```python
from transformers import pipeline
from datasets import load_dataset
from IPython.display import Audio, display
import torch
from scipy.io.wavfile import write

# Load model
model = pipeline("text-to-speech", model="microsoft/speecht5_tts")

# Load speaker voice
data = load_dataset("Matthijs/cmu-arctic-xvectors", split="validation")
speaker = torch.tensor(data[0]["xvector"]).unsqueeze(0)

# Input text
text = "Hello everyone, welcome to my presentation."

# Generate speech
result = model(text, forward_params={"speaker_embeddings": speaker})

# Save and play audio
audio = result["audio"].squeeze()
write("speech.wav", result["sampling_rate"], audio)
display(Audio("speech.wav"))

print("Audio generated successfully!")
```
