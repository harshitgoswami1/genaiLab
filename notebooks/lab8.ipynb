```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 1. Dataset
texts = [
    "The product is excellent",
    "The service is very bad",
    "I love this phone",
    "The movie was boring"
]

actual = ["positive", "negative", "positive", "negative"]

# 2. Zero-shot prompting
def zero_shot(text):
    positive = ["excellent", "love", "good"]
    negative = ["bad", "boring", "terrible"]

    text = text.lower()

    if any(word in text for word in positive):
        return "positive"
    return "negative"

# 3. Few-shot prompting
def few_shot(text):
    examples = {
        "I like this": "positive",
        "This is terrible": "negative"
    }

    for sentence, label in examples.items():
        if sentence.lower() in text.lower():
            return label

    return zero_shot(text)

# 4. Role-based prompting
def role_based(text):
    # Act as a sentiment analysis expert
    return zero_shot(text)

# 5. Structured reasoning
def reasoning(text):
    words = text.lower().split()
    positive = ["excellent", "love", "good"]
    negative = ["bad", "boring", "terrible"]

    if any(word in words for word in positive):
        return "positive"
    elif any(word in words for word in negative):
        return "negative"
    return "negative"

# 6. Structured output
def structured_output(text):
    result = reasoning(text)
    return {"text": text, "sentiment": result}

# 7. Display structured output
predicted = []

print("STRUCTURED OUTPUT:")
for text in texts:
    result = structured_output(text)
    print(result)
    predicted.append(result["sentiment"])

# 8. Evaluation
print("\nEVALUATION:")
print("Accuracy:", accuracy_score(actual, predicted))
print("Precision:", precision_score(actual, predicted,
                                    pos_label="positive"))
print("Recall:", recall_score(actual, predicted,
                              pos_label="positive"))
print("F1 Score:", f1_score(actual, predicted,
                            pos_label="positive"))
```
