```python
from transformers import pipeline

en_hi = pipeline("translation", model="Helsinki-NLP/opus-mt-en-hi")
en_fr = pipeline("translation", model="Helsinki-NLP/opus-mt-en-fr")
en_de = pipeline("translation", model="Helsinki-NLP/opus-mt-en-de")
```

```python
sentences = [
    "Artificial intelligence is transforming the world.",
    "India is a beautiful country.",
    "I love learning new things.",
    "The weather is very nice today.",
    "Education is important for everyone.",
    "Technology makes our lives easier.",
    "I am going to college.",
    "She is reading a book.",
    "He plays football every day.",
    "We should protect the environment.",
    "My family lives in India.",
    "Water is essential for life.",
    "The sun rises in the east.",
    "I want to learn a new language.",
    "Hard work leads to success.",
    "The train is arriving soon.",
    "Please open the window.",
    "Today is a beautiful day.",
    "Students are preparing for exams.",
    "Generative AI can create new content."
]
```

```python
english = []
hindi = []
french = []
german = []

for sentence in sentences:
    english.append(sentence)
    hindi.append(en_hi(sentence)[0]["translation_text"])
    french.append(en_fr(sentence)[0]["translation_text"])
    german.append(en_de(sentence)[0]["translation_text"])

print("Translation completed!")
```

```python
import pandas as pd

df = pd.DataFrame({
    "English": english,
    "Hindi": hindi,
    "French": french,
    "German": german
})

df.head()
```

```python
import evaluate

bleu = evaluate.load("sacrebleu")

# Example reference translations for the first sentence.
# Replace these with verified human translations for your dataset.
references_hi = ["कलापूर्ण बुद्धि दुनिया बदल है।"]
references_fr = ["L'intelligence artificiele transforme le monde."]
references_de = ["Künstliche Intelligenz verändert die Welt."]

bleu_hi = bleu.compute(predictions=[hindi[0]], references=[references_hi])
bleu_fr = bleu.compute(predictions=[french[0]], references=[references_fr])
bleu_de = bleu.compute(predictions=[german[0]], references=[references_de])
print(f"BLEU score for Hindi: {bleu_hi['score']:.2f}")
print(f"BLEU score for French: {bleu_fr['score']:.2f}")
print(f"BLEU score for German: {bleu_de['score']:.2f}") 
```
