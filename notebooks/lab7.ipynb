```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Create documents
docs = [
    "SolarOne is a subsidiary of GreenPower Ltd.",
    "GreenPower Ltd operates in the renewable energy industry.",
    "TechSoft develops computer software."
]

# Step 2: Create a retriever
vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(docs)

def retrieve(query):
    q = vectorizer.transform([query])
    scores = cosine_similarity(q, vectors)
    index = scores.argmax()
    return docs[index]

# Step 3: Ask a question
question = "What industry does SolarOne's parent company operate in?"

# Hop 1: Find the parent company
fact1 = retrieve("SolarOne subsidiary parent company")
print("Hop 1:", fact1)

# Extract the parent company
parent = fact1.split("subsidiary of ")[1].rstrip(".")

# Hop 2: Find the parent's industry
fact2 = retrieve(parent + " industry")
print("Hop 2:", fact2)

# Step 4: Generate the answer
print("Final Answer:", fact2)
```
