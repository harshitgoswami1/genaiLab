```python
import numpy as np
import pandas as pd

sentence = "The cat sat on the mat"
tokens = sentence.split()
n = len(tokens)

np.random.seed(42)
scores = np.random.rand(n, n)

print("Tokens:", tokens)

print("\nBefore masking:")
print(pd.DataFrame(scores, index=tokens, columns=tokens).round(2))

mask = np.tril(np.ones((n, n), dtype=int))

print("\nCausal mask (1 = allowed, 0 = blocked):")
print(mask)

masked_scores = np.where(mask == 1, scores, 0)

print("\nAfter masking:")
print(pd.DataFrame(masked_scores, index=tokens, columns=tokens).round(2))
```
